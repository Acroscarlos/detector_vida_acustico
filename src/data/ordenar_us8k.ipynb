```python
# ==========================================
# SCRIPT DE EXTRACCIÓN Y AGRUPACIÓN DE URBANSOUND8K
# ==========================================
import os
import pandas as pd
import shutil

# Rutas basadas en tu estructura de directorios
BASE_DIR = r"C:\Users\carlo\Documents\detector_vida_acustico\detector_vida_acustico"
US8K_AUDIO_DIR = os.path.join(BASE_DIR, "dataset_original", "UrbanSound8K", "audio")
US8K_CSV_PATH = os.path.join(BASE_DIR, "dataset_original", "UrbanSound8K", "metadata", "UrbanSound8K.csv")

# Carpeta temporal de clasificación (Staging Area)
# Aquí se organizarán antes de que tú las audites y las pases a 'dataset_preprocesado'
OUTPUT_DIR = os.path.join(BASE_DIR, "dataset_original", "UrbanSound8K_Organizado")

def agrupar_urbansound():
    print("⚙️ [SISTEMA] Leyendo metadata de UrbanSound8K.csv...")
    
    try:
        df = pd.read_csv(US8K_CSV_PATH)
    except Exception as e:
        print(f"❌ [ERROR] No se pudo leer el CSV: {e}")
        return

    # 1. Crear las 10 carpetas correspondientes a las clases originales
    clases_us8k = df['class'].unique()
    print(f"📂 Creando carpetas para las clases: {', '.join(clases_us8k)}")
    
    for clase in clases_us8k:
        ruta_clase = os.path.join(OUTPUT_DIR, str(clase))
        os.makedirs(ruta_clase, exist_ok=True)

    archivos_movidos = 0
    errores = 0

    print(f"📦 [PROCESANDO] Extrayendo audios de los folds y agrupando por clase...")
    
    # 2. Iterar sobre cada fila del CSV para ubicar el archivo exacto
    for index, row in df.iterrows():
        nombre_archivo = row['slice_file_name']
        carpeta_fold = f"fold{row['fold']}"
        nombre_clase = row['class']
        
        ruta_origen = os.path.join(US8K_AUDIO_DIR, carpeta_fold, nombre_archivo)
        ruta_destino = os.path.join(OUTPUT_DIR, nombre_clase, nombre_archivo)
        
        # 3. Copiar el archivo (usamos copy2 para no destruir el dataset original)
        if os.path.exists(ruta_origen):
            shutil.copy2(ruta_origen, ruta_destino)
            archivos_movidos += 1
        else:
            errores += 1

    print("\n" + "="*50)
    print("✅ [EXTRACCIÓN COMPLETADA]")
    print(f"▶ Archivos copiados y organizados: {archivos_movidos}")
    if errores > 0:
        print(f"⚠️ Archivos no encontrados en origen: {errores}")
    print("="*50)
    
    print("\n🚨 [DIRECTIVA DE AUDITORÍA MANUAL]:")
    print("1. Revisa las carpetas 'children_playing' y 'street_music' meticulosamente. Contienen voces.")
    print("2. Pasa los audios validados a tus 5 carpetas maestras dentro de 'dataset_preprocesado'.")

if __name__ == "__main__":
    agrupar_urbansound()
```
